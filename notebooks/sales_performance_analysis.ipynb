```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries loaded successfully")
```

```python
# Create sample sales dataset

data = {
    "Order_Date": ["2024-01","2024-02","2024-03","2024-04","2024-05","2024-06"],
    "Region": ["North","South","East","West","North","South"],
    "Category": ["Technology","Furniture","Office Supplies","Technology","Furniture","Office Supplies"],
    "Sales": [12000,9000,7500,15000,11000,8000]
}

sales_df = pd.DataFrame(data)

sales_df
```

```python
sales_df.info()
sales_df.describe()
```

```python
plt.figure(figsize=(8,5))
sns.barplot(x="Region", y="Sales", data=sales_df)
plt.title("Sales by Region")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.barplot(x="Category", y="Sales", data=sales_df)
plt.title("Sales by Product Category")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.lineplot(x="Order_Date", y="Sales", data=sales_df, marker="o")
plt.title("Monthly Sales Trend")
plt.show()
```
